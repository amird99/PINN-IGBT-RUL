```python
import matplotlib.pyplot as plt

series = build_cycle_series(cfg.data)
fig, ax = plt.subplots(figsize=(7, 3.5))
for name, x in series.items():
    ax.plot(x, label=name)
ax.set(xlabel='Cycle', ylabel='Mean V_CE (V)')
ax.legend(); plt.show()
```
